```python
n=int(input("enter the number:"))
temp=n
rev=0
count=len(str(n))
for i in range(count):
    digit=temp % 10
    rev=rev*10+digit
    temp=temp // 10
if n==rev:
    print("palindrome")
else:
    print("Not palindrome")
```
